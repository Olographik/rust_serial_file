```python
from rust.rust_python_file import ExamplStruct, PyExampleFile, PyPosition
import os
```

```python
f = PyExampleFile("test.file", False)
s = ExamplStruct(120)
f.write(s)

s = f.read_at(0)
if s is None:
    raise ValueError("Nothing to read")
print(s.index)

print(f.item_count())

f.write_at(PyPosition(1), ExamplStruct(12))
print(f.item_count())

s = f.read_at(0)
if s is None:
    raise ValueError("Nothing to read")
print(s.index)

s = f.read_at(1)
if s is None:
    raise ValueError("Nothing to read")
print(s.index)


os.remove("test.file")
```
